### k-means++
What about being able to choose the centers that are kind of far away from each other?

Let $\mathcal X$ be the input dataset of points. Suppose that we wnt to compute a clustering of $\mathcal X$ into $k \in \mathbb N$ clusters. Then, the $\texttt{k-means++}$ algorithm can be used to compute a set $\mathcal C = \{\mu_1, \ldots, \mu_k\}$ of centroids to be provided as initialization to the the Lloyd's algorithm:
1. $\mu_1 \leftarrow$ point $x \in \mathcal X$ chosen uniformly at random;
2. $\mathcal C \leftarrow \set{\mu_1}$;
3. for $i \leftarrow 2$ to $k$ do:
   - $\mu_i \leftarrow$ point $x \in \mathcal X \setminus \mathcal C$ chosen at random with probability $\Pr(x \mid \mathcal C)$;
   - $\mathcal C \leftarrow \mathcal C \cup \set{\mu_i}$;
4. return $\mathcal C$;

where

$\Pr(x \mid \mathcal C) = \frac{d(x, \mathcal C)^2}{\sum_{x' \in \mathcal X \setminus \mathcal C} (d(x', \mathcal C))^2}$

assigns a probability to a point of being selected as the next centroid proportional to its distance from the centroids collected in the previous iterations.




```python
# TO COMPLETE

# cluster the input data with KMeans, initializing centroids with k-means++
y_pred = KMeans(n_clusters=3, init='k-means++', n_init=1, random_state=numero_di_matricola+2).fit_predict(X)

# plot points colored based on the predicted clustering
plt.scatter(X[:, 0], X[:, 1], c=y_pred)
```

## Choice of the number k of clusters

Lloyd's algorithm assumes that the number $k$ of desired clusters is passed as input. In some applications it makes sense, but in others we do not know $k$. How can decide the value of $k$?

In general, we can use the following apporach:
1. cluster the input dataset $\mathcal X$ with different values of $k \in \mathbb N$, producing a clustering $\mathcal C_k$ for each value of $k$ considered;
2. compute some score $s(\mathcal C_k)$ to evaluate each clustering $\mathcal C_k$;
3. choose the value of $k$ that produces the clustering $\mathcal C_k$ that optimizes the score $s(\mathcal C_k)$.

What should we use as score function $s(\cdot)$?

### Objective Function Minimization

Let's see how the value of the objective function changes when using $k \in \{2, 3, 4, 5\}$. For each value of $k$, we also _print the value of the objective function_ (have a look at the documentation of scikit-learn).

```python
# TO COMPLETE

# use KMeans with k=2 to cluster the input data and print the value of the objective function (see the documentation)
clustering_k2 = KMeans(n_clusters=2, init='k-means++', n_init=5, random_state=numero_di_matricola).fit(X)
print(clustering_k2.inertia_)

# use KMeans with k=3 to cluster the input data and print the value of the objective function (see the documentation)
clustering_k3 = KMeans(n_clusters=3, init='k-means++', n_init=5, random_state=numero_di_matricola).fit(X)
print(clustering_k3.inertia_)

# use KMeans with k=4 to cluster the input data and print the value of the objective function (see the documentation)
clustering_k4 = KMeans(n_clusters=4, init='k-means++', n_init=5, random_state=numero_di_matricola).fit(X)
print(clustering_k4.inertia_)

# use KMeans with k=5 to cluster the input data and print the value of the objective function (see the documentation)
clustering_k5 = KMeans(n_clusters=5, init='k-means++', n_init=5, random_state=numero_di_matricola).fit(X)
print(clustering_k5.inertia_)
```

### Silhouette Score

Let's use a different approach: the silhouette coefficient.
The silhouette coefficient is a very commonly used score to evaluate a clustering.

Consider a clustering $\mathcal C_k = \set{C_1, \ldots, C_k}$ of the input dataset $\mathcal X$ into $k$ clusters.
Let $x \in \mathcal X$ be a point in the dataset and let $C_{\sigma(x)}$ be the clustering assigned to $x$, where $\sigma: \mathcal X \to \set{1, \ldots, k}$.
Define:
$$A(x) = \frac{\sum_{x' \in C_{\sigma(x)}, x' \neq x} d(x, x')}{|C_{\sigma(x)})| - 1},$$

where $d(x, x')$ is the distance between $x$ and $x'$.

Define also:
$$B(x) = \min_{C_i \in \mathcal C_k, C_i \neq C_{\sigma(x)}} d(x, C_i),$$

where $d(x, C_i) = \frac{\sum_{x' \in C_i} d(x, x')}{|C_i|}$.

Then, the silhouette of a point $x \in \mathcal X$ is defined as:
$$s(x) = \frac{B(x) - A(x)}{\max \set{A(x), B(x)}}.$$

The intuition is that $s(x)$ measures if $x$ is closer to the points in the clusters it is assigned to than to the points in its closest cluster.

**Question**: what is the range for $s(x)$?

**Answer**: $-1 \leq s(x) \leq 1$, where:
- $s(x)$ close to $0$ indicates that $x$ is in the boundary between two clusters;
- $s(x)$ close to $1$ indicates that $x$ is much closer to the points in the cluster $x$ is assigned to than the points in its closest cluster (so the assignment of $x$ is very good);
- $s(x)$ close to $-1$ indicates that $x$ is much farther to the points in the cluster $x$ is assigned to than the points in its closest cluster (so the assignment of $x$ is very bad).

Finally, to measure the quality of an entire clustering $\mathcal C_k$, the average silhouette is computed as:
$$s(\mathcal C_k) = \frac{\sum_{x \in \mathcal X} s(x)}{|\mathcal X|}.$$

**To do next**: for each of the clusterings obtained above (with $k \in \{2,3,4,5\}$), compute its average silhouette score (again, see the documentation of scikit-learn for more details).

```python
# TO COMPLETE

# import the silhouette_score function from sklearn.metrics
from sklearn.metrics import silhouette_score

# compute and print the average silhouette_score for the clustering computed with k=2
silhouette_avg_k2 = silhouette_score(X, clustering_k2.predict(X))
print("The average silhouette_score for k=2 is :",silhouette_avg_k2)

# compute and print the average silhouette_score for the clustering computed with k=3
silhouette_avg_k3 = silhouette_score(X, clustering_k3.predict(X))
print("The average silhouette_score for k=3 is :",silhouette_avg_k3)

# compute and print the average silhouette_score for the clustering computed with k=4
silhouette_avg_k4 = silhouette_score(X, clustering_k4.predict(X))
print("The average silhouette_score for k=4 is :",silhouette_avg_k4)

# compute and print the average silhouette_score for the clustering computed with k=5
silhouette_avg_k5 = silhouette_score(X, clustering_k5.predict(X))
print("The average silhouette_score for k=5 is :",silhouette_avg_k5)
```

According to the silhouette score, which is the best value for $k$? Let's also have a look at the objective function.

```python
# print the value of the objective function for each computed clustering
print(clustering_k2.inertia_)
print(clustering_k3.inertia_)
print(clustering_k4.inertia_)
print(clustering_k5.inertia_)
```

By looking both at the value of the objective function and at the silhouette score, which is the best value for $k$? Why?

Let's plot the (normalized) objective function and the silhouette score to better understand what happens.

```python
n_clusters = range(2, 11)
silhouettes = []
obj_fun_scores = []

for k in n_clusters:
    clustering = KMeans(n_clusters=k, 
                        init='k-means++',
                        n_init=1, 
                        random_state=numero_di_matricola).fit(X)
    silhouette = silhouette_score(X, clustering.predict(X))
    silhouettes.append(silhouette)
    obj_fun_scores.append(clustering.inertia_)
    print(f"Clustering with k={k}: Cost={clustering.inertia_:.2f}, Silhouette={silhouette:.2f}")

obj_fun_scores = np.array(obj_fun_scores)
obj_fun_scores = obj_fun_scores / max(obj_fun_scores)

plt.plot(n_clusters, silhouettes, label='Silhouette')
plt.plot(n_clusters, obj_fun_scores, label='Norm. Cost')
plt.title(r'Scores vs $k$')
plt.xlabel(r'$k$')
plt.ylabel('Score')
plt.legend();
```

### Agglomerative Clustering
Let's try with a different clustering technique: _agglomerative_ clustering!

Given an input dataset $\mathcal X$, agglomerative clustering produces a clustering $\mathcal C$ with $k$ clusters as follows:
1. $\mathcal C \leftarrow \set{\set{x} : x \in \mathcal X}$ (i.e., assign each point $x \in \mathcal X$ to a different cluster that contains only $x$);
2. while $|\mathcal C| > k$ do:
   - $(C_i, C_j) \leftarrow \arg\min_{C_i, C_j \in \mathcal C} d(C_i, C_j)$ (i.e., find the closest clusters $C_i, C_j \in \mathcal C$);
   - $\mathcal C \leftarrow \mathcal C \setminus \set{C_i, C_j} \cup \set{C_i \cup C_j}$ (i.e., replace $C_i$ and $C_j$ with their union);
3. return $\mathcal C$;

The function $d(C_i, C_j)$ measures the distance between the clusters $C_i, C_j \in \mathcal C$ and sometimes is called _linkage_. In scikit-learn you can choose among several options for $d(\cdot)$. Look at the documentation to find them.


```python
# TO COMPLETE

# cluster the input dataset using agglomerative clustering with single linkage (look at the documentation)
y_pred = AgglomerativeClustering(n_clusters=2, linkage='single').fit_predict(X)
plt.scatter(X[:, 0], X[:, 1], c=y_pred)
```

```python
import random

def christmas_tree():
    tree_height = 16

    print("\n\n🎄 Merry Christmas! 🎄\n".center(40))

    for i in range(tree_height):
        
        spaces = " " * (tree_height - i - 1)

        if i == 0:
            layer_content = "⭐"
        else:
            contents = [ "🍃", "🟡", "🔴", "🔵"]
            content = random.choices(contents, weights = [0.7, 0.1, 0.1, 0.1], k = 2 * i + 1)
            delimiter = ""
            layer_content = delimiter.join(content)

        print((spaces + layer_content).center(40))

    trunk = " " * (tree_height - 1) + "🟫"
    print(trunk.center(40))

christmas_tree()
```
